```python
# ==== Osztálykönyvtárak telepítése ====
import os, shutil
from PIL import Image
import matplotlib.pyplot as plt

# ==== SEGÍTSÉG SZÖVEGEK ====
resize_help = """A resize parancs használata:
A parancson kívül két paramétert vár: új szélesség és magasság
Pl.: resize 300 200
"""
crop_help = """A crop parancs használata:
A parancson kívül négy paramétert vár: a vágási téglalap két szemközti csúcsának koordinátái
Pl.: crop 10 5 160 250
"""
black_help = """A black parancs használata:
A parancson kívül nem vár paramétert.
Pl.: black
"""
rotate_help = """A rotate parancs használata:
A parancson kívül egy paramétert vár: a forgatás mértéke fokban (előjelesen)
Pl.: rotate -120
"""
```

```python
def scripthelp():
    print(" KÉPSZERKESZTŐ PROGRAM - Készítette: Kirilla József-FWN10Y)")
    print("--------------------------------------------")
    print("Elérhető parancsok:")
    print("  resize <szélesség> <magasság>   - Kép átméretezése")
    print("  crop <x1> <y1> <x2> <y2>        - Kép kivágása adott koordináták alapján")
    print("  rotate <fok>                    - Kép forgatása fokban (pozitív = óramutató járásával megegyezően)")
    print("  black                           - Kép fekete-fehérré alakítása")
    print("  help                            - Ez a helper megjelenítése")
    print("--------------------------------------------")
    print("Jegyzet: a módosított képek a /content/<mappa> könyvtárban találhatók (pl. /resized, /bw, /rotated)\n")
    print(resize_help)
    print(crop_help)
    print(rotate_help)
    print(black_help)
```

```python
def resize(newWidth, newHeight):
    """Képek átméretezése /content mappában"""
    items = os.listdir('/content')
    out_dir = '/content/resized'

    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.mkdir(out_dir)

    found = False
    for each_image in items:
        if each_image.lower().endswith((".jpg", ".png")):
            found = True
            full_path = os.path.join("/content", each_image)
            save_path = os.path.join(out_dir, f"resized_{newWidth}x{newHeight}_{each_image}")
            image = Image.open(full_path)
            image = image.resize((int(newWidth), int(newHeight)), Image.LANCZOS)
            image.save(save_path)
            print(full_path + " resized successfully!")

    if not found:
        print("Nincs szerkeszthető kép a /content mappában!")
```

```python
def crop(cropAx, cropAy, cropBx, cropBy):
    """Képek kivágása adott koordináták alapján"""
    items = os.listdir('/content')
    out_dir = '/content/cropped'

    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.mkdir(out_dir)

    found = False
    for each_image in items:
        if each_image.lower().endswith((".jpg", ".png")):
            found = True
            full_path = os.path.join("/content", each_image)
            save_path = os.path.join(out_dir, f"cropped_{each_image}")
            image = Image.open(full_path)
            image = image.crop((int(cropAx), int(cropAy), int(cropBx), int(cropBy)))
            image.save(save_path)
            print(full_path + " cropped successfully!")

    if not found:
        print(" Nincs szerkeszthető kép a /content mappában!")
```

```python
def rotate(degree):
    """Képek forgatása fokban"""
    items = os.listdir('/content')
    out_dir = '/content/rotated'

    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.mkdir(out_dir)

    found = False
    for each_image in items:
        if each_image.lower().endswith((".jpg", ".png")):
            found = True
            full_path = os.path.join("/content", each_image)
            save_path = os.path.join(out_dir, f"rotated_{degree}_{each_image}")
            image = Image.open(full_path)
            image = image.rotate(int(degree))
            image.save(save_path)
            print(full_path + " rotated successfully!")

    if not found:
        print(" Nincs szerkeszthető kép a /content mappában!")
```

```python
def black():
    """Képek fekete-fehérré alakítása"""
    items = os.listdir('/content')
    out_dir = '/content/bw'

    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.mkdir(out_dir)

    found = False
    for each_image in items:
        if each_image.lower().endswith((".jpg", ".png")):
            found = True
            full_path = os.path.join("/content", each_image)
            save_path = os.path.join(out_dir, f"bw_{each_image}")
            image = Image.open(full_path)
            image = image.convert("L").convert("1")
            image.save(save_path)
            print(full_path + " black-whited successfully!")

    if not found:
        print(" Nincs szerkeszthető kép a /content mappában!")
```

```python
command = input("Adj meg egy parancsot (pl.: resize 300 200, rotate 90, black, help):\n> ")
args = command.strip().split()

if not args:
    print(" Nem adtál meg parancsot. Írj be 'help'-et a súgóhoz.")
else:
    cmd = args[0].lower()

    try:
        if cmd == "resize":
            resize(args[1], args[2])
        elif cmd == "crop":
            crop(args[1], args[2], args[3], args[4])
        elif cmd == "rotate":
            rotate(args[1])
        elif cmd == "black":
            black()
        elif cmd in ["help", "-h", "info"]:
            scripthelp()
        else:
            print(" Ismeretlen parancs! Írj be 'help'-et a súgóhoz.")
    except Exception as e:
        print(" Hiba történt:", e)
```
