```python
# ==== Osztálykönyvtárak telepítése ====
import os, shutil
from PIL import Image
import matplotlib.pyplot as plt

# ==== SEGÍTSÉG SZÖVEGEK ====
resize_help = """A resize parancs használata:
A parancson kívül két paramétert vár: új szélesség és magasság
Pl.: resize 300 200
"""
crop_help = """A crop parancs használata:
A parancson kívül négy paramétert vár: a vágási téglalap két szemközti csúcsának koordinátái
Pl.: crop 10 5 160 250
"""
black_help = """A black parancs használata:
A parancson kívül nem vár paramétert.
Pl.: black
"""
rotate_help = """A rotate parancs használata:
A parancson kívül egy paramétert vár: a forgatás mértéke fokban (előjelesen)
Pl.: rotate -120
"""
```

```python
def resize(newWidth, newHeight):
    """Képek átméretezése /content mappában"""
    items = os.listdir('/content')
    out_dir = '/content/resized'

    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.mkdir(out_dir)

    found = False
    for each_image in items:
        if each_image.lower().endswith((".jpg", ".png")):
            found = True
            full_path = os.path.join("/content", each_image)
            save_path = os.path.join(out_dir, f"resized_{newWidth}x{newHeight}_{each_image}")
            image = Image.open(full_path)
            image = image.resize((int(newWidth), int(newHeight)), Image.LANCZOS)
            image.save(save_path)
            print(full_path + " resized successfully!")

    if not found:
        print("Nincs szerkeszthető kép a /content mappában!")
```
