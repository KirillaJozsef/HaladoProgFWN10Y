```python
import os, shutil, random # Fájlkezelés, mappák törlése, véletlenszerű műveletek
from PIL import Image, ImageEnhance, ImageOps # Képfeldolgozás, fényerő/kontraszt és tükrözés
import matplotlib.pyplot as plt # Képek megjelenítése
from google.colab import files  # Fájlok feltöltése Colab-ba

# Súgószövegek a parancsokhoz
resize_help = """resize <szélesség> <magasság>   - Kép átméretezése"""
crop_help   = """crop <x1> <y1> <x2> <y2>        - Kép kivágása"""
black_help  = """black                           - Fekete-fehér kép"""
rotate_help = """rotate <fok>                    - Kép forgatása fokban"""
label_help  = """label <címke> <képnév>          - Címke hozzárendelése egy képhez"""
augment_help = """augment                         - Adatnövelés (forgatás, tükrözés stb.)"""

# Súgó függvény kiíratása
def scripthelp():
    print(" KÉPSZERKESZTŐ PROGRAM - Kiegészítve labeling és augmentation funkcióval")
    print("---------------------------------------------------")
    print(resize_help)
    print(crop_help)
    print(rotate_help)
    print(black_help)
    print(label_help)
    print(augment_help)
    print("---------------------------------------------------")
    print("Használat: a parancsokat az alábbi módon írd:")
    print("  resize 300 200")
    print("  rotate 90")
    print("  crop 10 5 160 250")
    print("  label kutya dog.jpg")
    print("  augment")
    print("---------------------------------------------------")
```

```python
def label(label_name, image_name):  #Hozzárendeli a megadott címkét a kiválasztott képhez, és elmenti egy fájlba
    """Címke (label) hozzárendelése egy képhez"""
    label_file = "/content/labels.txt"
    if not os.path.exists(os.path.join("/content", image_name)):
        print(f"Hiba: a {image_name} nevű kép nem található!")
        return
    with open(label_file, "a") as f:
        f.write(f"{image_name}: {label_name}\n")
    print(f"Címke elmentve: {image_name} -> {label_name}")
```

```python
def augment():
    """Mesterséges adatnövelés: forgatás, tükrözés, fényerő, kontraszt"""
    items = os.listdir('/content')
    out_dir = '/content/augmented'

    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.mkdir(out_dir)

    found = False
    for each_image in items:
        if each_image.lower().endswith((".jpg", ".png")):
            found = True
            full_path = os.path.join("/content", each_image)
            img = Image.open(full_path)

            # 1. Forgatás véletlenszerűen
            angle = random.choice([-30, -15, 15, 30])
            img_rot = img.rotate(angle)
            img_rot.save(os.path.join(out_dir, f"rotated_{angle}_{each_image}"))

            # 2. Tükrözés
            img_flip = ImageOps.mirror(img)
            img_flip.save(os.path.join(out_dir, f"flipped_{each_image}"))

            # 3. Fényerő / kontraszt változtatás
            enhancer = ImageEnhance.Contrast(img)
            img_contrast = enhancer.enhance(random.uniform(0.7, 1.5))
            img_contrast.save(os.path.join(out_dir, f"contrast_{each_image}"))

            print(f"{each_image} augmentálva!")
    if not found:
        print("Nincs szerkeszthető kép a /content mappában!")
```

```python
"""
Ez a program a /content mappában található képeket dolgozza fel.
- resize: képek átméretezése a megadott szélességre és magasságra (/content/resized)
- crop: képek kivágása megadott koordináták szerint (/content/cropped)
- rotate: képek elforgatása megadott fokkal (/content/rotated)
- black: képek fekete-fehérré alakítása (/content/bw)
Minden függvény ellenőrzi, hogy van-e szerkeszthető kép,
és minden feldolgozott képről visszajelzést ad.
"""

def resize(newWidth, newHeight):
    items = os.listdir('/content')
    out_dir = '/content/resized'
    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.mkdir(out_dir)
    found = False
    for each_image in items:
        if each_image.lower().endswith((".jpg", ".png")):
            found = True
            full_path = os.path.join("/content", each_image)
            save_path = os.path.join(out_dir, f"resized_{newWidth}x{newHeight}_{each_image}")
            image = Image.open(full_path)
            image = image.resize((int(newWidth), int(newHeight)), Image.LANCZOS)
            image.save(save_path)
            print(full_path + " resized successfully!")
    if not found:
        print("Nincs szerkeszthető kép a /content mappában!")

def crop(cropAx, cropAy, cropBx, cropBy):
    items = os.listdir('/content')
    out_dir = '/content/cropped'
    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.mkdir(out_dir)
    found = False
    for each_image in items:
        if each_image.lower().endswith((".jpg", ".png")):
            found = True
            full_path = os.path.join("/content", each_image)
            save_path = os.path.join(out_dir, f"cropped_{each_image}")
            image = Image.open(full_path)
            image = image.crop((int(cropAx), int(cropAy), int(cropBx), int(cropBy)))
            image.save(save_path)
            print(full_path + " cropped successfully!")
    if not found:
        print("Nincs szerkeszthető kép a /content mappában!")

def rotate(degree):
    items = os.listdir('/content')
    out_dir = '/content/rotated'
    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.mkdir(out_dir)
    found = False
    for each_image in items:
        if each_image.lower().endswith((".jpg", ".png")):
            found = True
            full_path = os.path.join("/content", each_image)
            save_path = os.path.join(out_dir, f"rotated_{degree}_{each_image}")
            image = Image.open(full_path)
            image = image.rotate(int(degree))
            image.save(save_path)
            print(full_path + " rotated successfully!")
    if not found:
        print("Nincs szerkeszthető kép a /content mappában!")

def black():
    items = os.listdir('/content')
    out_dir = '/content/bw'
    if os.path.exists(out_dir):
        shutil.rmtree(out_dir)
    os.mkdir(out_dir)
    found = False
    for each_image in items:
        if each_image.lower().endswith((".jpg", ".png")):
            found = True
            full_path = os.path.join("/content", each_image)
            save_path = os.path.join(out_dir, f"bw_{each_image}")
            image = Image.open(full_path)
            image = image.convert("L").convert("1")
            image.save(save_path)
            print(full_path + " black-whited successfully!")
    if not found:
        print("Nincs szerkeszthető kép a /content mappában!")
```

```python
command = input("Adj meg egy parancsot (pl.: resize 300 200, rotate 90, crop 10 5 160 250, black, label kutya dog.jpg, augment, help):\n> ")
args = command.strip().split()

if not args:
    print("Nem adtál meg parancsot. Írj be 'help'-et a súgóhoz.")
else:
    cmd = args[0].lower()
    try:
        if cmd == "resize":
            resize(args[1], args[2])
        elif cmd == "crop":
            crop(args[1], args[2], args[3], args[4])
        elif cmd == "rotate":
            rotate(args[1])
        elif cmd == "black":
            black()
        elif cmd == "label":
            label(args[1], args[2])
        elif cmd == "augment":
            augment()
        elif cmd in ["help", "-h", "info"]:
            scripthelp()
        else:
            print("Ismeretlen parancs! Írj be 'help'-et a súgóhoz.")
    except Exception as e:
        print("Hiba történt:", e)
```
